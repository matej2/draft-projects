```python
from anthropic import Anthropic
client = Anthropic()
model = "claude-sonnet-4-0"
```

```python
def add_user_message(messages, text):
    user_message = {
        "role": "user",
        "content": text
    }
    messages.append(user_message)

def add_assistant_message(messages, text):
    assistant_message = {
        "role": "assistant",
        "content": text
    }
    messages.append(assistant_message)


def chat(messages):
    message = client.messages.create(
        model=model,
        max_tokens=1000,
        messages=[
            {
                "role": "user",
                "content": "What is quantum computing? Answer in one sentense"
            }
        ]
    )

    return message.content[0].text
```

```python
messages = []

add_user_message(messages, "Define quantum computing in one sentense.")

answer = chat(messages)

add_assistant_message(messages, answer)

add_user_message(messages, "Write another sentense")
```
